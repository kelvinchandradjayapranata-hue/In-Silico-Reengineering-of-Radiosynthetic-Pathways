# Docking AutoDock Vina di Google Colab (CPU)

Notebook ini menjalankan docking DHN monomer/dimer/trimer terhadap reseptor
surrogate PKS1/laccase. **Gunakan runtime CPU** (Runtime > Change runtime type > CPU);
Vina tidak memakai GPU dan CPU lokal tidak perlu dibebani.

Hasil ditulis ke `results/docking_results.json` dan `results/docking_summary.tsv`.
Pose/log per seed diarsipkan di `results/colab_runs/<run_id>/`.

Jika repositori masih privat, simpan token GitHub read-only sebagai Colab
Secret bernama `GITHUB_TOKEN`. Sel clone di bawah memakainya tanpa menulis
token ke `.git/config`.

```python
import os
import subprocess

REPO_URL = "https://github.com/kelvinchandradjayapranata-hue/In-Silico-Reengineering-of-Radiosynthetic-Pathways.git"
DEST = "/content/In-Silico-Reengineering-of-Radiosynthetic-Pathways"

try:
    from google.colab import userdata
    token = userdata.get('GITHUB_TOKEN')
except Exception:
    token = None

clone_url = REPO_URL
if token:
    clone_url = REPO_URL.replace(
        'https://github.com/',
        f'https://x-access-token:{token}@github.com/',
    )
subprocess.run(['git', 'clone', clone_url, DEST], check=False)
subprocess.run(['git', '-C', DEST, 'remote', 'set-url', 'origin', REPO_URL], check=False)
os.chdir(DEST)
print('Working directory:', os.getcwd())
```
